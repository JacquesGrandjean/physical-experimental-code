```python
import matplotlib.pyplot as plt
from scipy.signal import find_peaks
import numpy as np

mz = []
intensity = []

file_path = "spectra_mass_alkeller_CaM__eb5b6dd0.jdx"

with open(file_path, "r") as f:
    read_data = False
    
    for line in f:
        line = line.strip()
        
        if line.startswith("##PEAK TABLE"):
            read_data = True
            continue
        
        if read_data:
            if line.startswith("##"):
                break
            
            parts = line.split()
            if len(parts) == 2:
                mz.append(float(parts[0]))
                intensity.append(float(parts[1]))

mz = np.array(mz)
intensity = np.array(intensity)

# Détection des pics
peaks, _ = find_peaks(intensity, height=max(intensity)*0.2)  # seuil 20%

plt.figure(figsize=(10,5))
plt.plot(mz, intensity)

# Annoter les pics
for p in peaks:
    plt.text(mz[p], intensity[p], f"{mz[p]:.2f}", 
             rotation=0,
             verticalalignment='bottom',
             horizontalalignment='center')
plt.xlim(19000, 20000)
plt.xlabel("m [Da]")
plt.ylabel("Relative Intensity")
plt.title("Mass Spectrum: CaM")
plt.show()
```
